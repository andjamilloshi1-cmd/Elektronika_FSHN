# 2. Baza Teorike

## 2.1 Modeli i Qarkut

Pjesëtuesi i tensionit ka dy rezistenca, \(R_1\) dhe \(R_2\), të lidhura njëra pas tjetrës (në seri). Në hyrje vendoset tensioni \(V_{in}\), ndërsa tensioni në dalje \(V_{out}\) matet në pikën ku bashkohen dy rezistencat, kundrejt tokëzimit (GND).

Kur lidhim rezistencën e ngarkesës \(R_L\), ajo lidhet paralelisht me \(R_2\). Kjo ndryshon rezistencën e përgjithshme të pjesës së poshtme të qarkut dhe, si pasojë, mund të ndryshojë edhe tensionin në dalje.

Në llogaritjet teorike supozojmë se burimi dhe komponentët punojnë normalisht. Në matjet reale mund të ketë dallime të vogla për shkak të tolerancës së rezistencave, instrumenteve matëse dhe lidhjeve.

## 2.2 Ekuacionet Themelore

**Pjesëtuesi i tensionit pa ngarkesë:**

\[
V_{out}=V_{in}\frac{R_2}{R_1+R_2}
\]

**Rezistenca ekuivalente kur ngarkesa lidhet në paralel me \(R_2\):**

\[
R_{eq}=R_2\parallel R_L=\frac{R_2R_L}{R_2+R_L}
\]

**Tensioni dalës me ngarkesë:**

\[
V_{out}=V_{in}\frac{R_{eq}}{R_1+R_{eq}}
\]

**Raporti i tensioneve:**

\[
A_v=\frac{V_{out}}{V_{in}}
\]

Në këtë eksperiment raporti \(A_v\) përdoret si raport i pjesëtimit të tensionit dhe jo si amplifikim aktiv.

# 5. Analiza e të Dhënave

## 5.1 Analiza e Eksperimentit 1

Përdorim formulën e pjesëtuesit të tensionit:

\[
V_{out(teorik)}=15,35\cdot\frac{7,78}{7,55+7,78}
\]

\[
V_{out(teorik)}\approx 7,79\,V
\]

Tensioni i matur është \(7,50\,V\). Diferenca absolute është:

\[
\Delta V=|7,50-7,79|\approx 0,29\,V
\]

Gabimi relativ, duke marrë si referencë vlerën teorike, është:

\[
\varepsilon=\frac{|V_{out(matur)}-V_{out(teorik)}|}
{V_{out(teorik)}}\cdot100\%
\]

\[
\varepsilon\approx 3,72\%
\]

Pra, matja eksperimentale është afër vlerës teorike, me një diferencë të vogël.

## 5.2 Analiza e Eksperimentit 2

Raporti eksperimental i tensioneve është:

\[
A_{v(matur)}=\frac{6,71}{6,82}\approx 0,984
\]

Raporti teorik i përcaktuar nga rezistencat është:

\[
A_{v(teorik)}=\frac{3,30}{3,27+3,30}\approx 0,502
\]

Këto dy raporte nuk përputhen. Për më tepër, vlera 15,77 V është shënuar veçmas në fletën laboratorike, pa u identifikuar qartë si tension hyrës apo si një matje tjetër. Prandaj, kjo pjesë kërkon verifikimin e lidhjeve dhe të etiketave të matjeve përpara se të nxirret një përfundim sasior përfundimtar.

## 5.3 Analiza e Eksperimentit 3

Për secilën matje llogarisim rezistencën ekuivalente \(R_{eq}\).

**Matja 1:**

\[
R_{eq1}=\frac{7,78\cdot 3300}{7,78+3300}\,k\Omega
\]

\[
R_{eq1}\approx 7,762\,k\Omega
\]

**Matja 2:**

\[
R_{eq2}=\frac{7,78\cdot 2340}{7,78+2340}\,k\Omega
\]

\[
R_{eq2}\approx 7,754\,k\Omega
\]

Rezistenca e ngarkesës është shumë më e madhe se \(R_2\), prandaj lidhja paralele e ndryshon shumë pak rezistencën ekuivalente të degës së poshtme.

Raportet teorike të pjesëtimit për dy rastet janë afërsisht:

\[
\frac{R_{eq1}}{R_1+R_{eq1}}\approx 0,5069
\]

\[
\frac{R_{eq2}}{R_1+R_{eq2}}\approx 0,5067
\]

Në fletën laboratorike nuk është shënuar \(V_{in}\) për këto dy matje. Për këtë arsye nuk mund të llogaritet një vlerë numerike e plotë e \(V_{out(teorik)}\), por mund të krahasohen rezistencat ekuivalente dhe raportet teorike të pjesëtimit. Tensionet dalëse të matura janë përkatësisht 7,76 V dhe 7,71 V.

```python
# Llogaritjet numerike të eksperimenteve
R1_1, R2_1, Vin_1, Vout_matur_1 = 7.55, 7.78, 15.35, 7.50
Vout_teorik_1 = Vin_1 * R2_1 / (R1_1 + R2_1)
gabimi_1 = abs(Vout_matur_1 - Vout_teorik_1) / Vout_teorik_1 * 100

R1_2, R2_2 = 3.27, 3.30  # MΩ
Vin_2, Vout_matur_2 = 6.82, 6.71
raporti_teorik_2 = R2_2 / (R1_2 + R2_2)
raporti_matur_2 = Vout_matur_2 / Vin_2

R1_3, R2_3 = 7.55, 7.78  # kΩ
ngarkesat_Mohm = [3.30, 2.34]
rezultate = []
for RL_Mohm in ngarkesat_Mohm:
    RL_kohm = RL_Mohm * 1000
    Req = (R2_3 * RL_kohm) / (R2_3 + RL_kohm)
    raport = Req / (R1_3 + Req)
    rezultate.append((RL_Mohm, Req, raport))

print(f"Eksperimenti 1: Vout teorik = {Vout_teorik_1:.3f} V")
print(f"Gabimi relativ i eksperimentit 1 = {gabimi_1:.2f}%")
print(f"Eksperimenti 2: raport teorik = {raporti_teorik_2:.4f}")
print(f"Eksperimenti 2: raport i matur = {raporti_matur_2:.4f}")
for i, (RL, Req, raport) in enumerate(rezultate, start=1):
    print(f"Matja {i}: RL = {RL:.2f} MΩ, Req = {Req:.3f} kΩ, raporti teorik = {raport:.4f}")
```
